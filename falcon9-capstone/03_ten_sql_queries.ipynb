# All ten SQL tasks
Original reproducible analysis of the retained IBM course snapshots. Execute from the repository root. Python source files and datasets accompany these notebooks. Historical cohort definitions and unavailable live metadata are explicit. No physical launch or new live API acquisition is claimed.

```python
import sqlite3,pandas as pd,json
from IPython.display import display
connection=sqlite3.connect("capstone-checks/spacex.db")
```

## Task 1: Unique launch sites

```python
query = 'SELECT DISTINCT Launch_Site FROM SPACEXTBL ORDER BY Launch_Site'
result = pd.read_sql_query(query,connection)
display(result)
```

## Task 2: Five CCA launch records

```python
query = "SELECT Date, Booster_Version, Launch_Site, PAYLOAD_MASS__KG_ FROM SPACEXTBL WHERE Launch_Site LIKE 'CCA%' LIMIT 5"
result = pd.read_sql_query(query,connection)
display(result)
```

## Task 3: NASA CRS payload mass

```python
query = "SELECT SUM(PAYLOAD_MASS__KG_) AS Total_kg FROM SPACEXTBL WHERE Customer='NASA (CRS)'"
result = pd.read_sql_query(query,connection)
display(result)
```

## Task 4: F9 v1.1 average payload

```python
query = "SELECT AVG(PAYLOAD_MASS__KG_) AS Average_kg FROM SPACEXTBL WHERE Booster_Version LIKE 'F9 v1.1%'"
result = pd.read_sql_query(query,connection)
display(result)
```

## Task 5: First successful ground landing

```python
query = "SELECT MIN(Date) AS First_date FROM SPACEXTBL WHERE Landing_Outcome='Success (ground pad)'"
result = pd.read_sql_query(query,connection)
display(result)
```

## Task 6: Drone-ship boosters, 4000-6000 kg

```python
query = "SELECT DISTINCT Booster_Version FROM SPACEXTBL WHERE Landing_Outcome='Success (drone ship)' AND PAYLOAD_MASS__KG_>4000 AND PAYLOAD_MASS__KG_<6000 ORDER BY Booster_Version"
result = pd.read_sql_query(query,connection)
display(result)
```

## Task 7: Mission outcomes

```python
query = 'SELECT TRIM(Mission_Outcome) AS Mission_Outcome, COUNT(*) AS Missions FROM SPACEXTBL GROUP BY TRIM(Mission_Outcome) ORDER BY Missions DESC'
result = pd.read_sql_query(query,connection)
display(result)
```

## Task 8: Boosters with maximum payload

```python
query = 'SELECT Booster_Version, PAYLOAD_MASS__KG_ FROM SPACEXTBL WHERE PAYLOAD_MASS__KG_=(SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTBL) ORDER BY Booster_Version'
result = pd.read_sql_query(query,connection)
display(result)
```

## Task 9: Drone-ship landing failures in 2015

```python
query = "SELECT substr(Date,6,2) AS Month, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTBL WHERE substr(Date,1,4)='2015' AND Landing_Outcome='Failure (drone ship)' ORDER BY Date"
result = pd.read_sql_query(query,connection)
display(result)
```

## Task 10: Landing outcomes within date interval

```python
query = "SELECT Landing_Outcome, COUNT(*) AS Outcomes FROM SPACEXTBL WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY Outcomes DESC"
result = pd.read_sql_query(query,connection)
display(result)
```
